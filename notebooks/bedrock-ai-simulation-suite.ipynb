```python
from groq import Groq
import os
import random
import hashlib

client = Groq(api_key=os.getenv("GROQ_API_KEY"))

def fake_llm(prompt: str):
    """Simula una llamada LLM usando Groq."""
    completion = client.chat.completions.create(
        model="groq/compound-mini",
        messages=[{"role": "user", "content": prompt}],
        temperature=0.3,
        max_tokens=300
    )
    return completion.choices[0].message.content
```

## 1️⃣ Prompt Engineering avanzado

Definimos un prompt de sistema con ejemplos y roles.


```python
system_prompt = """
You are a helpful assistant that summarizes product reviews.
Here are examples:

Review: The new phone has great battery life but poor camera quality.
Summary: Reviewer praises battery life but criticizes camera quality.

Review: The laptop is fast and lightweight, perfect for travel.
Summary: Reviewer highlights speed and portability.
"""

user_input = "The smartwatch looks stylish but the battery drains quickly."
prompt = system_prompt + "\nReview: " + user_input + "\nSummary:"
print(fake_llm(prompt))
```

## 2️⃣ Caché de respuestas

Simulamos almacenamiento en caché para reducir tokens y acelerar respuestas.


```python
CACHE = {}

def cache_key(system, user):
    return hashlib.sha256((system + user).encode()).hexdigest()

key = cache_key(system_prompt, user_input)

if key in CACHE:
    print("[Cache Hit]", CACHE[key])
else:
    result = fake_llm(prompt)
    CACHE[key] = result
    print("[Cache Miss]", result)
```

## 3️⃣ Evaluación automática y modelo juez

```python
def automatic_eval(response: str):
    return {
        "coherence": random.uniform(0.7, 1.0),
        "clarity": random.uniform(0.7, 1.0),
        "length": len(response.split())
    }

response = fake_llm(prompt)
metrics = automatic_eval(response)
print("Metrics:", metrics)

def judge_eval(prompt, response):
    judge_prompt = f"Rate the following response from 1 to 10.\nPrompt: {prompt}\nResponse: {response}\nGive only the number."
    return fake_llm(judge_prompt)

print("Judge Score:", judge_eval(prompt, response))
```

## 4️⃣ Recuperación Aumentada (RAG)

```python
documents = [
    "Smartwatches track health metrics like heart rate and sleep.",
    "Battery optimization is key for wearable devices.",
    "Design and comfort influence user adoption."
]

def rag_retrieve(query):
    return [doc for doc in documents if any(word in doc.lower() for word in query.lower().split())]

def rag_generate(query, retrieved_docs):
    context = " ".join(retrieved_docs)
    prompt = f"Using this context: {context}\nAnswer: {query}"
    return fake_llm(prompt)

query = "How can smartwatch battery life be improved?"
retrieved = rag_retrieve(query)
generated = rag_generate(query, retrieved)

print("Retrieved:", retrieved)
print("Generated:", generated)
```

## 5️⃣ Procesamiento multimedia simulado

```python
def simulate_transcription():
    prompt = "Simulate a short transcription of a person talking about technology trends."
    return fake_llm(prompt)

transcription = simulate_transcription()
summary = fake_llm(f"Summarize this transcription:\n{transcription}")

logos = random.sample(["Apple", "Samsung", "Google", "Microsoft"], k=random.randint(0, 2))
iab_tags = random.sample(["Technology", "Lifestyle", "Health", "Education"], k=2)

analysis = {
    "transcription": transcription,
    "summary": summary,
    "logos_detected": logos,
    "iab_tags": iab_tags
}

analysis
```

## 6️⃣ Enrutamiento inteligente de modelos

```python
def route_model(prompt):
    if len(prompt.split()) < 10:
        return "groq/compound-mini"
    else:
        return "groq/llama-3.1-70b-versatile"

chosen_model = route_model(prompt)
print("Chosen Model:", chosen_model)
```
